```python
import pandas as pd
import matplotlib as plt
import csv
import psycopg
from sqlalchemy import create_engine
import os
```

```python
from dotenv import load_dotenv
load_dotenv("../../../.env")
```

```python
engine = create_engine(
    f"postgresql+psycopg://{os.getenv('DB_USER')}:{os.getenv('DB_PASSWORD')}@{os.getenv('DB_HOST')}:{os.getenv('DB_PORT')}/{os.getenv('DB_NAME')}"
)
```

```python
accounts_df = pd.read_sql_query('SELECT * from accounts;', engine)
exchanges_df = pd.read_sql_query('SELECT * from exchanges;', engine)
holdings_df = pd.read_sql_query('SELECT * from holdings;', engine)
instruments_df = pd.read_sql_query('SELECT * from instruments;', engine)
trades_df = pd.read_sql_query('SELECT * from trades;', engine)
user_pii_df = pd.read_sql_query('SELECT * from user_pii;', engine)
users_df = pd.read_sql_query('SELECT * from users;', engine)
```
